### Exercício 2
Implementar o método iterativo geral que aparece no quadro laranja acima e testar ele para os casos de método de Jacobi e Gauss-Seidel. Comparar com os resultados anteriores

```python
import numpy as np
from tabulate import tabulate

def ij2n (i, j, N):
    return i + j*N

def MatAssembly(N): #monta a matriz com índices de 0 até N^2 -1
    nunk = N*N;
    A = np.zeros(shape=(nunk,nunk))
    for i in range(1,N-1):
        for j in range(1,N-1):
            # aqui estão sendo definidos os índices de cada nó 
            Ic = ij2n(i, j, N)      #central
            Ie = ij2n(i+1, j, N)    #leste
            Iw = ij2n(i-1, j, N)    #oeste
            In = ij2n(i, j+1, N)    #norte
            Is = ij2n(i, j-1, N)    #sul
            
            # aqui preenche a linha do central --> média dos quatros extremos
            A[Ic,[Ic,Ie,Iw,In,Is]] = [4.0, -1.0, -1.0, -1.0, -1.0]
    return A

def BuildSystem(N, TL, TR, TB, TT):
    A = MatAssembly(N)
    Atilde = A.copy()
    
    nunk = N**2
    b = np.zeros(shape=(nunk,1)) 
    k = np.array(range(0,N)) 
    Iden = np.identity(nunk) 

    Ic = ij2n(0,k,N)
    Atilde[Ic,:], b[Ic] = Iden[Ic,:], TL #
    
    Ic = ij2n(N-1,k,N)
    Atilde[Ic,:], b[Ic] = Iden[Ic,:], TR 
    
    Ic = ij2n(k,0,N)
    Atilde[Ic,:], b[Ic] = Iden[Ic,:], TB 

    Ic = ij2n(k,N-1,N)
    Atilde[Ic,:], b[Ic] = Iden[Ic,:], TT 
    
    return Atilde, b

def MetIntGeral(A, b, x0, TOL, MAX_IT, met): 
    # TOL é a tolerância do residual
    # MAX_IT é quantidade máxima de iterações permitidas
    # x0 é o chute inicial
    if met == 0:
        M = np.diag(np.diag(A))
    if met == 1:
        M = np.tril(A)
    
    r = A@x0 - b
    k = 0 
    while(np.linalg.norm(r) > TOL and k < MAX_IT):
        d = np.linalg.solve(M,-r)
        beta = 1
        x = x0 + beta*d
        r = A@x -b
        x0 = x
        k += 1
    return k

MAX_IT = np.array([200, 500, 1000])
TOL = np.array([10**(-5), 10**(-8)])
N = np.array([11,21,31])
TL = 0.0
TR = 0.0
TB = 0.0
TT = 20.0
jac = []
gs = []

for i in range(N.shape[0]):
    A, b = BuildSystem(N[i], TL, TR, TB, TT)
    x0 = np.full((A.shape[0], 1), 5)
    res_jac = MetIntGeral(A, b, x0, TOL[0], MAX_IT[i], 0)
    res_gs = MetIntGeral(A, b, x0, TOL[0], MAX_IT[i], 1)
    jac.append(res_jac)
    gs.append(res_gs)

dados1 = [
    ["N", "TOL", "Iterações Jacobi", "Iterações Gauss-Seidel"],
    [N[0], TOL[0], jac[0], gs[0]],
    [N[1], TOL[0], jac[1], gs[1]],
    [N[2], TOL[0], jac[2], gs[2]]
]
print("Tabela 1:\n", tabulate(dados1))

for i in range(N.shape[0]):
    A, b = BuildSystem(N[i], TL, TR, TB, TT)
    x0 = np.full((A.shape[0], 1), 5)
    res_jac = MetIntGeral(A, b, x0, TOL[1], MAX_IT[i], 0)
    res_gs = MetIntGeral(A, b, x0, TOL[1], MAX_IT[i], 1)
    jac.append(res_jac)
    gs.append(res_gs)
    
dados2 = [
    ["N", "TOL", "Iterações Jacobi", "Iterações Gauss-Seidel"],
    [N[0], TOL[1], jac[0], gs[0]],
    [N[1], TOL[1], jac[1], gs[1]],
    [N[2], TOL[1], jac[2], gs[2]]
]

print("\nTabela 2:\n", tabulate(dados2))
```

### Exercício 3

Modificar a função que fixa as condições de contorno de forma
tal que se mantenha a simetria de matriz. Testar aplicando o
método iterativo geral usado anteriormente.

```python
import numpy as np
from tabulate import tabulate


def ij2n (i, j, N):
    return i + j*N


def MatAssembly(N): #monta a matriz com índices de 0 até N^2 -1
    nunk = N*N;
    A = np.zeros(shape=(nunk,nunk))
    for i in range(1,N-1):
        for j in range(1,N-1):
            # aqui estão sendo definidos os índices de cada nó 
            Ic = ij2n(i, j, N)      #central
            Ie = ij2n(i+1, j, N)    #leste
            Iw = ij2n(i-1, j, N)    #oeste
            In = ij2n(i, j+1, N)    #norte
            Is = ij2n(i, j-1, N)    #sul
            
            # aqui preenche a linha do central --> média dos quatros extremos
            A[Ic,[Ic,Ie,Iw,In,Is]] = [4.0, -1.0, -1.0, -1.0, -1.0]
    return A


# Função para modificar a matriz e incorporar as condições de borda
def BuildSystemSim(N, TL, TR, TB, TT):
    nunk = N**2
    A = MatAssembly(N)
    Atilde = A.copy()
    b = np.zeros(shape=(nunk))
    k = np.array(range(0,N))
    Iden = np.identity(nunk)

    Ic = ij2n(0,k,N) # Tleft
    for r in k:
        b[:] = b[:] - A[:,Ic[r]]*TL 
    Atilde[Ic,:], b[Ic] = Iden[Ic,:], TL
    Atilde[:, Ic] = (Atilde[Ic, :]).T

    Ic = ij2n(N-1,k,N)
    for r in k:
        b[:] = b[:] - A[:, Ic[r]]*TR
    Atilde[Ic,:], b[Ic] = Iden[Ic,:], TR # Tright
    Atilde[:, Ic] = (Atilde[Ic, :]).T

    Ic = ij2n(k,0,N)
    for r in k:
        b[:] = b[:] - A[:, Ic[r]]*TB
    Atilde[Ic,:], b[Ic] = Iden[Ic,:], TB # Tbottom
    Atilde[:, Ic] = (Atilde[Ic, :]).T

    Ic = ij2n(k,N-1,N)
    for r in k:
        b[:] = b[:] - A[:, Ic[r]]*TT
    Atilde[Ic,:], b[Ic] = Iden[Ic,:], TT # Ttop
    Atilde[:, Ic] = (Atilde[Ic, :]).T
    
    return Atilde, b


def MetIntGeral(A, b, x0, TOL, MAX_IT, met): 
    # TOL é a tolerância do residual
    # MAX_IT é quantidade máxima de iterações permitidas
    # x0 é o chute inicial e sempre muda para o x anterior
    if met == 0:
        M = np.diag(np.diag(A))
    if met == 1:
        M = np.tril(A)
    
    r = A@x0 - b
    k = 0 
    while(np.linalg.norm(r) > TOL and k < MAX_IT):
        d = np.linalg.solve(M,-r)
        beta = 1
        x = x0 + beta*d
        r = A@x -b
        x0 = x
        k += 1
    return k


MAX_IT = np.array([400, 1500, 2000])
TOL = np.array([10**(-5), 10**(-8)])
N = np.array([11, 21])
TL = 0.0
TR = 0.0
TB = 0.0
TT = 20.0
jac = []
gs = []

for i in range(N.shape[0]):
    A, b = BuildSystemSim(N[i], TL, TR, TB, TT)
    x0 = np.full((A.shape[0], 1), 5)
    res_jac = MetIntGeral(A, b, x0, TOL[0], MAX_IT[i], 0)
    res_gs = MetIntGeral(A, b, x0, TOL[0], MAX_IT[i], 1)
    jac.append(res_jac)
    gs.append(res_gs)

dados1 = [
    ["N", "TOL", "Iterações Jacobi", "Iterações Gauss-Seidel"],
    [N[0], TOL[0], jac[0], gs[0]],
    [N[1], TOL[0], jac[1], gs[1]],
    ##[N[2], TOL[0], jac[2], gs[2]],
]
print("Tabela 1:\n", tabulate(dados1))

for i in range(N.shape[0]):
    A, b = BuildSystemSim(N[i], TL, TR, TB, TT)
    x0 = np.full((A.shape[0], 1), 5)
    res_jac = MetIntGeral(A, b, x0, TOL[1], MAX_IT[i], 0)
    res_gs = MetIntGeral(A, b, x0, TOL[1], MAX_IT[i], 1)
    jac.append(res_jac)
    gs.append(res_gs)
    
dados2 = [
    ["N", "TOL", "Iterações Jacobi", "Iterações Gauss-Seidel"],
    [N[0], TOL[1], jac[0], gs[0]],
    [N[1], TOL[1], jac[1], gs[1]],
    #[N[2], TOL[0], jac[2], gs[2]],
]

print("\nTabela 2:\n", tabulate(dados2)) 


# É possível notar que, para matrizes simétricas, ambos os métodos precisam de mais iterações para conseguir
# atingir a solução do problema, sendo que não são sensíveis à mudança de tolerância
```

```python
import numpy as np
from tabulate import tabulate
import scipy
from scipy.linalg import solve 
import matplotlib.pyplot as plt

# Função para plotar os curvas de nível constante de temperatura
def PlotaTemperature(n,m,T):
    x = np.arange(0, n, 1)
    y = np.arange(0, m, 1)
    X,Y=np.meshgrid(x,y)
    Z = np.copy(T)
    fig, ax = plt.subplots(figsize=(6,6))
    ax.set_aspect('equal')
    ax.set(xlabel='x', ylabel='y', title='Contours of temperature')
    im = ax.contourf(X, Y, Z.T, 20, cmap='jet')
    im2 = ax.contour(X, Y, Z.T, 20, linewidths=0.25, colors='k')
    fig.colorbar(im, ax=ax)
    plt.show()
    

def ij2n (i, j, N):
    return i + j*N

def MatAssembly(N): #monta a matriz com índices de 0 até N^2 -1
    nunk = N*N;
    A = np.zeros(shape=(nunk,nunk))
    for i in range(1,N-1):
        for j in range(1,N-1):
            # aqui estão sendo definidos os índices de cada nó 
            Ic = ij2n(i, j, N)      #central
            Ie = ij2n(i+1, j, N)    #leste
            Iw = ij2n(i-1, j, N)    #oeste
            In = ij2n(i, j+1, N)    #norte
            Is = ij2n(i, j-1, N)    #sul
            
            # aqui preenche a linha do central --> média dos quatros extremos
            A[Ic,[Ic,Ie,Iw,In,Is]] = [4.0, -1.0, -1.0, -1.0, -1.0]
    return A

def BuildSystem(N, T, R):
    A = MatAssembly(N)
    Atilde = A.copy()
    
    nunk = N**2
    b = np.zeros(shape=(nunk,1)) # right-hand-side
    k = np.array(range(0,N)) # Auxiliary array
    Iden = np.identity(nunk) # Auxiliary matrix

    meio = int((N-1)/2)
    lado = 1/(N-1)
    qtd = int(R//lado)
    
    for i in range(N):
        for j in range(N):
            dist = (((i - meio) * lado) ** 2 + ((j - meio) * lado) ** 2) ** 0.5
            if dist <= 0.2:
                Ic = ij2n(i, j, N)
                Atilde[Ic, :], b[Ic] = Iden[Ic, :], T
            else:
                if i == 0 or i == N-1 or j == 0 or j == N-1:
                    Ic = ij2n(i, j, N)
                    Atilde[Ic, :], b[Ic] = Iden[Ic, :], 0
    
    return Atilde, b


N = np.array([11,31,71])

for i in range(N.shape[0]):
    A, b = BuildSystem(N[i], 20, 0.2)
    T = scipy.linalg.solve(A,b)
    Tplot = T.copy().reshape(N[i],N[i]).T
    PlotaTemperature(N[i],N[i],Tplot)
```

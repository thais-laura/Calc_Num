## 1.7 Gráficos e plots

### Gráficos 2D básicos

```python
import numpy as np
import matplotlib.pyplot as plt

xi = 0
xf = 2.0*np.pi
N = 1000

x = np.linspace(xi, xf, N)
y = 2.0 + np.sin(2.0*x)+ 0.05*np.cos(100.0*x)
y2 = 2.0 + np.cos(2.0*x) + 0.1*np.cos(20.0*x)
plt.plot(x,y, '-or', x, y2, 'g')
plt.grid()
plt.xlabel('t [sec]')
plt.ylabel('y(t)')
plt.title('Meu grafico')
plt.legend(['Função 1', 'Função 2'])
plt.show()
```

## 1.9 Subplots

```python
fig, ax = plt.subplots(2,2, figsize=(20,20))

y3 = np.exp(-0.5*x)
y4 = np.tanh(y2)

ax[0,0].plot(x,y, '-r')
ax[0,0].legend(['funcao 1'])
ax[0,0].grid()
ax[0,0].set_ylabel('y1(t)')

ax[0,1].plot(x,y2, '-og')
ax[1,0].plot(x,y3, '-y')
ax[1,1].plot(x,y4)
```

```python
from mpl_toolkits import mplot3d

def myfunc(x,y):
    return (1 - (x**2 + y**3))*np.exp(-(x**2+y**2)/2)

x = np.arange(-3.0,3.0,0.1)
y = np.arange(-3.0,3.0,0.1)
X,Y = np.meshgrid(x,y)
Z = myfunc(X, Y)

fig = plt.figure()
ax = plt.axes(projection = "3d", facecolor='w')
ax.plot_surface(X, Y, Z)
ax.set(xlabel="$x$", ylabel="$y$", zlabel="$f(x, y)$", title='$z=(1-x^2+y^3) e^{-{x^2+y^2}/2}$')
       
m = plt.cm.ScalarMappable(cmap = 'RdBu')
m.set_array(Z)
plt.show()
```

## 1.10 Histograma

```python
n = 3
N = 100
somas = [] #criando uma lista vazia
ti = time.time()
for i in range(N):
    x = np.random.rand(n)
    s = np.sum(x)
    somas.append(s)
    if i < 5:
        print(x)
        print(s)
        print(somas)
print(time.time() -ti)
plt.hist(somas)
```
